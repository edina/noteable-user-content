## 2. Structure the Notebook Like a Learning Journey

<div style="border:1px solid #c3e6cb; background:#d4edda; color:#155724; padding:16px; margin:8px 0; border-radius:4px;">
    <b>Teaching idea:</b> Good exemplars have narrative flow. Think of the notebook as a journey with a beginning, middle, and end rather than a pile of unrelated demonstrations.
</div>

A useful structure is:

- <b>Opening:</b> introduce the problem, scenario, or question
- <b>Setup:</b> load tools and explain what they are for
- <b>Core walkthrough:</b> build understanding step by step
- <b>Learner practice:</b> invite the user to change or extend something
- <b>Reflection:</b> summarise what was learned
- <b>Next steps:</b> suggest how the learner can continue

If possible, frame the notebook around a tiny story. For example:

- "You are analysing survey results for a student society"
- "You are exploring telescope data to identify a pattern"
- "You are cleaning a messy dataset before making a decision"
- "You are comparing two possible models for a real-world task"

Stories make notebooks feel purposeful. They help learners understand why each code step matters.

## 4. Show the Smallest Useful Example First

<div style="border:1px solid #c3e6cb; background:#d4edda; color:#155724; padding:16px; margin:8px 0; border-radius:4px;">
    <b>Teaching idea:</b> Start small. Before showing a complex workflow, give learners a tiny success they can understand immediately.
</div>

This is especially important in exemplars. Early success builds confidence.

```python
# A very small example you can adapt to your own subject.
# This demonstrates a basic pattern: input -> processing -> output.

def describe_value(value):
    doubled = value * 2
    print(f"Original value: {value}")
    print(f"Doubled value: {doubled}")

describe_value(5)
```

After your first mini-example, explain why it matters.

For instance, you might say:

- in data science, this pattern becomes <i>load -> transform -> analyse</i>
- in statistics, it becomes <i>input -> calculate -> interpret</i>
- in simulation, it becomes <i>parameters -> model -> output</i>
- in text analysis, it becomes <i>text -> processing -> insight</i>

This helps the learner connect the simple example to the bigger learning goal.

## 6. Explain the Why Between Code Cells

<div style="border:1px solid #c3e6cb; background:#d4edda; color:#155724; padding:16px; margin:8px 0; border-radius:4px;">
    <b>Teaching idea:</b> The spaces between code cells matter. Use markdown to explain not just what the code does, but why the learner is doing it now.
</div>

A weak notebook says: "Run this cell."

A stronger notebook says: "Run this cell to compare the categories and identify which one has the highest value. This prepares us for the visualisation in the next section."

```python
# Example of a small analysis step with meaningful output.

highest_row = example_data.loc[example_data["Value"].idxmax()]

print("The category with the highest value is:")
print(highest_row)
```

Notice how this kind of code cell can support explanation. You can now ask the learner:

- Were you expecting that result?
- What pattern do you think we will see in a chart?
- What might this mean in your scenario?

These small prompts make the notebook feel more conversational and reflective.

## 8. Visualise Results to Support Interpretation

<div style="border:1px solid #b8daff; background:#d9edf7; color:#0c5460; padding:16px; margin:8px 0; border-radius:4px;">
    <b>Goal:</b> Show how a visual can reinforce learning, not just decorate the notebook.
</div>

Visualisations should help answer a question. After every plot, tell the learner what to look for.

```python
print("Generating example visualisation...")

plt.figure(figsize=(6, 4))
plt.bar(example_data["Category"], example_data["Value"], color="#4C72B0")
plt.title("Example Data Overview")
plt.xlabel("Category")
plt.ylabel("Value")
plt.show()
```

Useful follow-up guidance after a chart might include:

- Which bar is tallest, and what does that imply?
- Do any categories look unusually low or high?
- Does the visual confirm what you expected from the table?
- What new question does this chart raise?

This is where notebooks begin to feel exploratory instead of purely procedural.

## 9. Add Simple Interactivity Where It Helps Learning

<div style="border:1px solid #c3e6cb; background:#d4edda; color:#155724; padding:16px; margin:8px 0; border-radius:4px;">
    <b>Interactivity advice:</b> Use widgets when they help the learner explore an idea. Do not add interactivity only because it looks impressive.
</div>

Good uses of interactivity include:

- changing a threshold or parameter
- comparing multiple scenarios
- exploring sensitivity
- seeing how output changes when input changes


```python
print("Building interactive example...")

def interactive_plot(multiplier):
    scaled_values = example_data["Value"] * multiplier
    
    plt.figure(figsize=(6, 4))
    plt.bar(example_data["Category"], scaled_values, color="#55A868")
    plt.title(f"Values scaled by {multiplier}")
    plt.xlabel("Category")
    plt.ylabel("Scaled value")
    plt.ylim(0, max(example_data["Value"]) * 5 + 5)
    plt.show()

multiplier_slider = widgets.IntSlider(
    value=1,
    min=1,
    max=5,
    step=1,
    description="Scale:",
    continuous_update=False
)

widgets.interact(interactive_plot, multiplier=multiplier_slider);
```

If you include interactivity, explain what learners should try. For example:

- Increase the multiplier and observe what stays the same
- Which features change, and which do not?
- What does this tell us about proportional scaling?

Without guidance, widgets can become decoration rather than learning tools.

Examples of effective notebook activities:

## 11. Include Helpful Error Guidance and Friction Points

<div style="border:1px solid #f5c6cb; background:#f8d7da; color:#721c24; padding:16px; margin:8px 0; border-radius:4px;">
    <b>Design caution:</b> A polished exemplar anticipates where learners may get stuck and offers reassurance, explanations, or checks.
</div>

Common issues to plan for:

- missing brackets or quotation marks
- cell execution out of order
- variables not yet defined
- file paths not matching the instructions
- using the wrong kernel or environment
- misunderstanding output rather than producing an actual code error

Rather than only showing ideal outcomes, consider adding short notes such as:

- "If you see a `NameError`, check whether you ran the setup cell first."
- "If your plot is blank, check the column names in the dataframe preview."
- "If the widget does not appear, ensure the cell has fully run."

This makes the notebook feel more supportive and realistic.

## 13. Exemplar Design Checklist

<div style="border:1px solid #c3e6cb; background:#d4edda; color:#155724; padding:16px; margin:8px 0; border-radius:4px;">
    <b>Final advice:</b> Before publishing your notebook, review it as if you were a learner seeing it for the first time.
</div>

Checklist:

- Does the title clearly say what this notebook is about?
- Does the introduction explain why the topic matters?
- Are the learning outcomes visible?
- Does each code cell have a purpose?
- Is there enough explanation between code cells?
- Does the notebook work from top to bottom without hidden assumptions?
- Are tasks manageable and meaningful?
- Are the outputs interesting or informative?
- Is there a narrative, question, or journey holding the notebook together?
- Have you checked accessibility, wording clarity, and consistency?
- Have you run every cell in order from a fresh kernel?
- Would a learner know what to do next after finishing?

If the answer to several of these is "not yet", that is normal. Exemplar notebooks improve significantly through small rounds of revision.
