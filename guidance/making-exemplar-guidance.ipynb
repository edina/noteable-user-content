## Narrative setup: why functions matter

Imagine you are writing Python code to help with small repeated tasks, such as:

- converting marks into percentages
- calculating an average
- checking whether a number is even
- generating a short message for a user

Without functions, you would repeat the same code again and again. Functions help us package useful logic into a named block that we can reuse.

## 1. A first tiny example

We will start with the smallest useful example.

Run the cell below. It defines a function called <code>greet</code>. The function takes one input, <code>name</code>, and prints a greeting.

```python
def greet(name):
    print(f"Hello, {name}!")

greet("Amina")
```

### What just happened?

- <code>def</code> starts a function definition
- <code>greet</code> is the function name
- <code>name</code> is a parameter, which means it is an input the function expects
- the indented line is the code that runs when the function is called
- <code>greet("Amina")</code> calls the function

This is a good example of a strong teaching pattern in a notebook: **show a tiny success first, then explain it clearly.**

## 2. Reusing a function with different inputs

A function becomes useful when we can call it more than once with different values.

```python
greet("Ben")
greet("Chloe")
greet("David")
```

We wrote the greeting logic once, but used it several times. That is one of the main reasons functions are valuable.

## 3. Functions that calculate and return values

Printing is useful, but many functions are more powerful when they **return** a value.

Run the next cell to define a function that doubles a number.

```python
def double(number):
    return number * 2

result = double(6)
print("The doubled value is:", result)
```

### Why use <code>return</code>?

A returned value can be stored in a variable, used in another calculation, or passed into another function.

That means functions help us build larger workflows from small parts.

```python
first_value = double(3)
second_value = double(10)

print("First value:", first_value)
print("Second value:", second_value)
print("Sum of both doubled values:", first_value + second_value)
```

## 4. A slightly more realistic example

Now let us move from a toy example to a small academic-style task.

Suppose a learner scored some marks and wants a simple function to convert a score into a percentage.

```python
def percentage(score, total):
    return (score / total) * 100

exam_percentage = percentage(45, 50)
print(f"Percentage: {exam_percentage:.1f}%")
```

This example is useful because it shows learners that functions are not just abstract syntax. They can encode a meaningful rule or process.

## 5. Understanding function inputs more clearly

The next function takes **two inputs** and returns their average.

```python
def average(a, b):
    return (a + b) / 2

print("Average of 10 and 14:", average(10, 14))
print("Average of 7 and 9:", average(7, 9))
```

When teaching with notebooks, this is a good point to pause and ask a learner to predict an output before they run the cell.

## 7. Functions can help organise repeated work

Suppose we want to classify whether a number is even or odd. A function makes that logic reusable.

```python
def even_or_odd(number):
    if number % 2 == 0:
        return "even"
    else:
        return "odd"

for value in [2, 5, 11, 20]:
    print(f"{value} is {even_or_odd(value)}")
```

This is a nice teaching moment because it combines:

- a function definition
- a conditional statement
- repeated function calls

It shows how notebook exemplars can gradually build complexity without becoming overwhelming.

## 8. Visualising outputs from a function

A notebook can also include a small visual to help learners interpret patterns.

Below, we will apply the <code>double</code> function to several values and plot the results.

```python
import matplotlib.pyplot as plt

inputs = [1, 2, 3, 4, 5]
outputs = [double(x) for x in inputs]

plt.figure(figsize=(6, 4))
plt.plot(inputs, outputs, marker="o")
plt.title("Input and output of the double function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid(True)
plt.show()
```

### What should learners notice?

The output grows in a straight-line pattern because each output is exactly twice the input.

A useful notebook does not just show a plot. It tells learners what to look for and why it matters.

## 9. Simple interactivity

Interactivity is most useful when it helps learners explore how changing an input affects an output.

In the example below, move the slider to see the effect of doubling different numbers.

```python
import ipywidgets as widgets
from IPython.display import display

def show_double(number):
    print(f"Input: {number}")
    print(f"Output from double(number): {double(number)}")

widgets.interact(show_double, number=widgets.IntSlider(value=5, min=0, max=20, step=1, description="Number"));
```

This interactivity is simple, but pedagogically useful. It reinforces the relationship between input and output.

## 10. Common pitfalls

When learners first work with functions, they often run into a few common issues:

- forgetting the colon at the end of the <code>def</code> line
- forgetting to indent the function body
- calling a function before the definition cell has been run
- printing when they meant to return, or returning when they meant to print
- using a variable name that does not exist
