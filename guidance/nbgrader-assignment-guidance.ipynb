## 2. Understanding nbgrader question types

<div style="border:1px solid #c3e6cb; background:#d4edda; color:#155724; padding:16px; margin:8px 0; border-radius:4px;">
    <b>Core idea:</b> nbgrader lets instructors mix different types of assessment in one notebook.
</div>

### The main types you will see
- **Manually graded answer**  
  A human grader reads your response and assigns marks.

- **Manually graded task**  
  A task description cell that tells you what to do across one or more cells.

- **Autograded answer**  
  You write code, and hidden tests check whether it behaves correctly.

- **Autograded tests**  
  Read-only test cells that run during grading.

- **Hidden tests**  
  Extra tests hidden from students so they cannot see every grading detail.

- **AUTOTEST / HASHED AUTOTEST**  
  Short syntax that lets nbgrader generate tests automatically.

## 8. Interactive Explorations

<div style="border:1px solid #c3e6cb; background:#d4edda; color:#155724; padding:16px; margin:8px 0; border-radius:4px;">
    <b>Interactivity:</b> Many libraries (like mapping or 3D tools) have interactivity built right in! For basic data, we can also use sliders to manipulate our graphs dynamically.
</div>

Run the cell below, and try using the controls to see how the output updates in real-time.

```python
print("Building interactive widget...")

def interactive_function(multiplier):
    # Scale the data based on the slider
    updated_values = my_data['Values'] * multiplier
    
    plt.figure(figsize=(6, 4))
    plt.bar(my_data['Category'], updated_values, color='#55A868')
    plt.title(f"Values Multiplied by {multiplier}")
    plt.xlabel("Category")
    plt.ylabel("Scaled Values")
    plt.ylim(0, 100) # Lock the Y-axis so it doesn't jump around
    plt.show()

# Create the interactive slider
my_widget = widgets.IntSlider(
    value=1,
    min=1,
    max=5,
    step=1,
    description='Multiplier:',
    continuous_update=False
)

print("Widget ready! Use the controls below:\n")

# Display the widget linked to our function
widgets.interact(interactive_function, multiplier=my_widget);
```
